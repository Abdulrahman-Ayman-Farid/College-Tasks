```python
# Abdulrahman Ayman Mohammad Farid Assignment 1
def vaccum_agent(percept):
    location, status = percept
    
    if status == 'Dirty':
        return 'Clean'
    elif location == 'A':
        return 'Right'
    else:  # Because location can only be 'A' or 'B'
        return 'Left'
    
print(vaccum_agent(('A', 'Dirty')))  # Output: Clean
print(vaccum_agent(('A', 'Clean')))  # Output: Right
```

```python
# List Comprehension example
percepts = [('A', 'Dirty'), ('B', 'Clean'), ('A', 'Clean')]
cleaned_actions = [vaccum_agent(percept) for percept in percepts]
print(cleaned_actions)  # Output: ['Clean', 'Left', 'Right']
```

```python
# Model-Based Vacuum Agent
class ModelBasedVacuumAgent:
    def __init__(self):
        self.model = { 'A' : None, 'B': None }

    def act(self, percept):
        location, status = percept
        self.model[location] = status
        
        if status == 'Dirty':
            return 'Clean'
        elif self.model['A'] == 'Clean' and self.model['B'] == 'Clean':
            return 'NoOp'
        return 'Right' if location == 'A' else 'Left'
    
agent = ModelBasedVacuumAgent()
print(agent.act(('A', 'Dirty')))  # Output: Clean
print(agent.act(('B', 'Clean')))  # Output: Left
```

```python
# Rational Agent Using PEAS
class PEAS:
    def __init__(self, performance, environment, actuators, sensors):
        self.performance = performance
        self.environment = environment
        self.actuators = actuators
        self.sensors = sensors

    def display(self):
        print("PEAS Description of the Agent for self-driving Taxi:")
        print(f"Performance: {self.performance}")
        print(f"Environment: {self.environment}")
        print(f"Actuators: {self.actuators}")
        print(f"Sensors: {self.sensors}")

taxi_peas = PEAS(
    performance="Maximize passenger satisfaction and safety",
    environment="Urban areas with varying traffic conditions",
    actuators="Steering, acceleration, braking, and signaling systems",
    sensors="Cameras, LIDAR, GPS, and ultrasonic sensors"
)
taxi_peas.display()
```

```python
# Search for Rational Path in Egypt Map

egypt_map = {  # I Changed Numbers for more diversity
    'Cairo': {'Giza': 10, 'Alexandria': 120},
    'Giza': {'Cairo': 10, 'Fayoum': 50},
    'Alexandria': {'Cairo': 120},
    'Fayoum': {'Giza': 50, 'Asyut': 100},
    'Asyut': {'Fayoum': 100, 'Luxor': 200},
    'Luxor': {'Asyut': 200, 'Hurghada': 150},
    'Hurghada': {'Luxor': 150, 'Sharm El Sheikh': 250},
    'Sharm El Sheikh': {'Hurghada': 250}
}
def total_cost(path):
    cost = 0
    for i in range(len(path) - 1):
        cost += egypt_map[path[i]][path[i + 1]]
    return cost

print("Total cost from Cairo to Sharm El Sheikh:", total_cost(['Cairo', 'Giza', 'Fayoum', 'Asyut', 'Luxor', 'Hurghada', 'Sharm El Sheikh']))  # Output: Total cost from Cairo to Sharm El Sheikh: 760
```

```python
# Agent with increasing autonomy
import random

class LearningAgent:
    def __init__(self):
        self.experience = {}

    def act(self, percept):
        if percept not in self.experience:
            self.experience[percept] = random.choice(['left', 'right', 'suck'])
        return self.experience[percept]

agent = LearningAgent()
print (agent.act(('A', 'Dirty')))  # Output: Random action from ['left', 'right', 'suck']
print (agent.act(('B', 'Clean')))  # Output: Random action from ['left', 'right', 'suck']
print (agent.experience)
```

```python
# Part Picking Robot: Action Based on camera Input

def part_picking_robot(part_type):
    if part_type == 'screw':
        return 'place in bin 1'
    elif part_type == 'bolt':
        return 'place in bin 2'
    else:
        return 'Reject part'
    
print(part_picking_robot('screw'))  # Output: place in bin 1
print(part_picking_robot('bolt'))   # Output: place in bin 2
print(part_picking_robot('nut'))    # Output: Reject part
```

```python
# Autonomy Level Simulation
class Agent:
    def __init__(self, name):
        self.name = name
        self.autonomy = 0.0

    def learn(self):
        self.autonomy += 0.1
    
agent = Agent("Crawling Robot")
for _ in range(10):
    agent.learn()
    print(f"{agent.name} autonomy level: {agent.autonomy:.1f}")  # Output: Crawling Robot autonomy level: 1.0
```
